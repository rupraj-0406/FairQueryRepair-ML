```python
def run_query_v2(data, q):

    result = data.copy()

    if q["age_op"] == ">=":
        result = result[result["age"] >= q["age_th"]]
    else:
        result = result[result["age"] <= q["age_th"]]

    if q["edu_op"] == ">=":
        result = result[result["education_num"] >= q["edu_th"]]
    else:
        result = result[result["education_num"] <= q["edu_th"]]

    if q["hours_op"] == ">=":
        result = result[result["hours_per_week"] >= q["hours_th"]]
    else:
        result = result[result["hours_per_week"] <= q["hours_th"]]

    result = result[
        result["capital_gain_binary"] == q["gain_val"]
    ]

    return result
```

```python
# Implementing SPD Calculation
def compute_spd(data):
    male = data[data["sex"] == "Male"]
    female = data[data["sex"] == "Female"]

    if len(male) == 0 or len(female) == 0:
        return None

    p_male = male["income"].mean()
    p_female = female["income"].mean()

    spd = p_male - p_female

    return spd
```

```python
def build_ml_repair(
    query,
    prediction
):

    repair = {}

    repair["age_op"] = decode_operator(
    query["age_op"]
    )

    repair["edu_op"] = decode_operator(
    query["edu_op"]
    )

    repair["hours_op"] = decode_operator(
    query["hours_op"]
    )

    repair["age_th"] = int(
        query["age_th"]
        + prediction["delta_age"]
    )

    repair["edu_th"] = int(
        query["edu_th"]
        + prediction["delta_edu"]
    )

    repair["hours_th"] = int(
        query["hours_th"]
        + prediction["delta_hours"]
    )

    repair["gain_val"] = int(
        query["gain_val"]
        + prediction["delta_gain"]
    )

    repair["gain_val"] = max(
        0,
        min(1, repair["gain_val"])
    )

    return repair
```

```python
def decode_operator(op):

    if op == 1:
        return ">="

    return "<="
```

```python
sample_query = X_test.iloc[0]

sample_pred = preds_df.iloc[0]

ml_repair = build_ml_repair(
    sample_query,
    sample_pred
)

print(ml_repair)
```

```python
from itertools import product

def generate_neighbors(
    ml_repair,
    radius=1
):

    age_vals = range(
        ml_repair["age_th"] - radius,
        ml_repair["age_th"] + radius + 1
    )

    edu_vals = range(
        ml_repair["edu_th"] - radius,
        ml_repair["edu_th"] + radius + 1
    )

    hours_vals = range(
        ml_repair["hours_th"] - radius,
        ml_repair["hours_th"] + radius + 1
    )

    gain_vals = [0, 1]

    neighbors = []

    for age, edu, hours, gain in product(
        age_vals,
        edu_vals,
        hours_vals,
        gain_vals
    ):

        candidate = ml_repair.copy()

        candidate["age_th"] = age
        candidate["edu_th"] = edu
        candidate["hours_th"] = hours
        candidate["gain_val"] = gain

        neighbors.append(candidate)

    return neighbors
```

```python
neighbors = generate_neighbors(
    ml_repair,
    radius=1
)

print(
    "Number of neighbors:",
    len(neighbors)
)

neighbors[:3]
```

```python
# Fairness Evaluator For Neighbors
def find_best_neighbor(
    neighbors,
    df
):

    fair_neighbors = []

    for candidate in neighbors:

        result = run_query_v2(
            df,
            candidate
        )

        if len(result) < 100:
            continue

        spd = compute_spd(result)

        if spd is None:
            continue

        if abs(spd) <= 0.20:

            fair_neighbors.append(
                (
                    candidate,
                    abs(spd),
                    len(result)
                )
            )

    return fair_neighbors
```

```python
guided_success = 0

guided_fail = 0

total_queries = len(X_test)

print(total_queries)
```

```python
for i in range(len(X_test)):

    query = X_test.iloc[i]

    prediction = preds_df.iloc[i]

    ml_repair = build_ml_repair(
        query,
        prediction
    )

    neighbors = generate_neighbors(
        ml_repair,
        radius=1
    )

    fair_neighbors = find_best_neighbor(
        neighbors,
        df
    )

    if len(fair_neighbors) > 0:

        guided_success += 1

    else:

        guided_fail += 1
```
